# Analyse der Leistung von zwei Photovoltaikanlagen

## Projektziel

In diesem Projekt werden die Produktions- und Wetterdaten von zwei Photovoltaikanlagen untersucht.

Ziel ist es herauszufinden, wie Sonneneinstrahlung und Temperatur die Stromproduktion beeinflussen. Außerdem wird untersucht, ob einzelne Wechselrichter oder Zeiträume ungewöhnlich niedrige Leistungen aufweisen.

## Hauptfrage

Wie beeinflussen Sonneneinstrahlung und Temperatur die Stromproduktion der beiden PV-Anlagen, und lassen sich auffällige Leistungseinbußen erkennen?

## Hypothesen

1. Je höher die Sonneneinstrahlung ist, desto höher ist die erzeugte AC-Leistung.
2. Die Modultemperatur hängt stärker mit der Stromproduktion zusammen als die Außentemperatur.
3. Einzelne Wechselrichter erzeugen weniger Energie als die übrigen Wechselrichter derselben Anlage.
4. Anlage 2 weist häufiger auffällige Leistungseinbrüche auf als Anlage 1.
5. Die höchste Stromproduktion wird ungefähr zur Mittagszeit erreicht.

## Datenqualität der DC-Werte prüfen

Vor der eigentlichen Analyse wird kontrolliert, wie sich die DC-Leistung und die AC-Leistung der beiden Anlagen zueinander verhalten.

Ein ungewöhnlich großes Verhältnis kann auf unterschiedliche Skalierungen oder Einheiten in den Ausgangsdaten hinweisen.

```python
# Gesamte AC- und DC-Leistung je Anlage berechnen

ac_dc_df = (
    analyse_df
    .groupby("anlage", as_index=False)
    .agg(
        ac_leistung_gesamt=("ac_leistung", "sum"),
        dc_leistung_gesamt=("dc_leistung", "sum")
    )
)

# Verhältnis von AC-Leistung zu DC-Leistung berechnen

ac_dc_df["ac_dc_verhaeltnis"] = (
    ac_dc_df["ac_leistung_gesamt"]
    / ac_dc_df["dc_leistung_gesamt"]
)

# Verhältnis als Prozentwert anzeigen

ac_dc_df["ac_dc_prozent"] = (
    ac_dc_df["ac_dc_verhaeltnis"] * 100
).round(1)

ac_dc_df
```

### Ergebnis der Datenqualitätsprüfung

Bei Anlage 1 beträgt das Verhältnis der gesamten AC-Leistung zur DC-Leistung nur ungefähr 9,8 %. Bei Anlage 2 liegt das Verhältnis bei ungefähr 97,8 %.

Die DC-Werte von Anlage 1 scheinen daher anders skaliert oder in einer anderen Einheit gespeichert zu sein. Die genaue Ursache kann mit den vorhandenen Informationen nicht bestimmt werden.

Für den Vergleich der beiden Anlagen werden deshalb hauptsächlich die AC-Leistung und relative Werte verwendet. Die absoluten DC-Werte der beiden Anlagen werden nicht direkt miteinander verglichen.

### Datenqualität: Anzahl der meldenden Wechselrichter

Die bisherige Berechnung verwendet die Summe der AC-Leistung aller Wechselrichter je Anlage und Zeitpunkt. Wenn einzelne Wechselrichter zu einem Zeitpunkt keine Messwerte liefern, fällt diese Summe niedriger aus. Dadurch kann ein Leistungseinbruch entstehen, obwohl lediglich Messwerte fehlen.

Bei Anlage 1 melden bei 99,7 % der untersuchten Tageszeitpunkte alle 22 Wechselrichter. Bei Anlage 2 trifft dies nur auf 70,7 % der Tageszeitpunkte zu.

Deshalb wurde zusätzlich die durchschnittliche AC-Leistung je meldendem Wechselrichter berechnet. Nach dieser Anpassung sinkt die Zahl der auffälligen Messzeitpunkte bei Anlage 2 von 30 auf 19. Bei Anlage 1 werden weiterhin keine Auffälligkeiten erkannt.

Die Korrelation zwischen Sonneneinstrahlung und AC-Leistung steigt bei Anlage 2 von 0,91 auf 0,93.

Die Kernaussage der Analyse bleibt damit bestehen: Anlage 2 weist auch nach Berücksichtigung der fehlenden Meldungen häufiger auffällige Leistungseinbrüche auf als Anlage 1. Die ursprüngliche Anzahl von 30 Messzeitpunkten wurde jedoch teilweise durch Datenlücken beeinflusst.

## Zusatzprüfung: Modultemperatur bei ähnlicher Sonneneinstrahlung

Die bisherige Korrelation zeigt, dass Modultemperatur und AC-Leistung gemeinsam ansteigen. Daraus kann jedoch nicht geschlossen werden, dass eine höhere Modultemperatur die Leistung erhöht.

Starke Sonneneinstrahlung erwärmt die Module und erhöht gleichzeitig die Stromproduktion. Um den Temperatureffekt besser zu untersuchen, werden deshalb nur Zeitpunkte mit ähnlicher Sonneneinstrahlung zwischen 0,5 und 0,8 betrachtet.

Zusätzlich werden nur Zeitpunkte verwendet, an denen alle 22 Wechselrichter Messwerte geliefert haben.

## Vergleichsdaten erstellen

```python
# Nur vollständige Zeitpunkte mit ähnlicher Einstrahlung auswählen

temperatur_pruefung_df = zeitpunkt_df.loc[
    (zeitpunkt_df["anzahl_wr"] == 22)
    & (zeitpunkt_df["sonneneinstrahlung"].between(0.5, 0.8))
].copy()

# Leistung je Wechselrichter und Einheit Einstrahlung berechnen

temperatur_pruefung_df["leistung_je_einstrahlung"] = (
    temperatur_pruefung_df["ac_je_wr"]
    / temperatur_pruefung_df["sonneneinstrahlung"]
)

# Modultemperaturen in Gruppen einteilen

temperatur_pruefung_df["temp_klasse"] = pd.cut(
    temperatur_pruefung_df["modultemperatur"],
    bins=[0, 40, 45, 50, 55, 80],
    labels=["unter 40", "40-45", "45-50", "50-55", "über 55"]
)

temperatur_pruefung_df.head()
```

## Durchschnitt nach Temperaturgruppe berechnen

```python
temperatur_vergleich_df = (
    temperatur_pruefung_df
    .groupby(
        ["anlage", "temp_klasse"],
        observed=True,
        as_index=False
    )
    .agg(
        durchschnitt_leistung_je_einstrahlung=(
            "leistung_je_einstrahlung", "mean"
        ),
        anzahl_messungen=(
            "leistung_je_einstrahlung", "count"
        )
    )
)

temperatur_vergleich_df[
    "durchschnitt_leistung_je_einstrahlung"
] = temperatur_vergleich_df[
    "durchschnitt_leistung_je_einstrahlung"
].round(0)

temperatur_vergleich_df
```

## Unterschied für Anlage 1 berechnen

```python
anlage1_temp = temperatur_vergleich_df.loc[
    temperatur_vergleich_df["anlage"] == "Anlage 1"
].set_index("temp_klasse")

leistung_unter_40 = anlage1_temp.loc[
    "unter 40",
    "durchschnitt_leistung_je_einstrahlung"
]

leistung_ueber_55 = anlage1_temp.loc[
    "über 55",
    "durchschnitt_leistung_je_einstrahlung"
]

leistung_ueber_55 = float(leistung_ueber_55)
leistung_unter_40 = float(leistung_unter_40)

# Prozentualen Unterschied berechnen und auf eine Nachkommastelle runden
unterschied_prozent = round(
    (leistung_ueber_55 - leistung_unter_40)
    / leistung_unter_40 * 100,1)

print(
    "Veränderung bei über 55 °C gegenüber unter 40 °C:",
    unterschied_prozent,
    "%"
)
```

## Analyse 4: Auffällige Leistungseinbrüche

### Hypothese

Anlage 2 weist häufiger auffällige Leistungseinbrüche auf als Anlage 1.

Als auffällig gilt in dieser Analyse ein Messzeitpunkt, bei dem die Sonneneinstrahlung mindestens 60 % des Höchstwertes der jeweiligen Anlage erreicht, während die AC-Leistung unter 30 % ihres Höchstwertes liegt.

## Werte normalisieren

```python
# AC-Leistung relativ zum Höchstwert jeder Anlage

zeitpunkt_df["ac_leistung_relativ"] = (zeitpunkt_df["ac_leistung"]/ zeitpunkt_df.groupby("anlage")["ac_leistung"].transform("max"))

zeitpunkt_df.head(3)
```

```python
# Sonneneinstrahlung relativ zum Höchstwert jeder Anlage berechnen

zeitpunkt_df["einstrahlung_relativ"] = (zeitpunkt_df["sonneneinstrahlung"] / zeitpunkt_df.groupby("anlage")["sonneneinstrahlung"].transform("max"))

zeitpunkt_df.head(3)
```

```python
# Messungen mit hoher Sonneneinstrahlung auswählen
hohe_einstrahlung_korrigiert_df = zeitpunkt_df.loc[
    zeitpunkt_df["einstrahlung_relativ"] >= 0.60
].copy()

# Leistungseinbruch mit der Leistung je meldendem Wechselrichter prüfen
hohe_einstrahlung_korrigiert_df["leistungseinbruch_korrigiert"] = (
    hohe_einstrahlung_korrigiert_df["ac_je_wr_relativ"] < 0.30
)

# Ergebnisse je Anlage zusammenfassen
leistungseinbrueche_korrigiert_df = (
    hohe_einstrahlung_korrigiert_df
    .groupby("anlage", as_index=False)
    .agg(
        messungen_bei_hoher_einstrahlung=(
            "leistungseinbruch_korrigiert", "count"
        ),
        anzahl_leistungseinbrueche=(
            "leistungseinbruch_korrigiert", "sum"
        )
    )
)

# Prozentualen Anteil berechnen
leistungseinbrueche_korrigiert_df["anteil_prozent"] = (
    leistungseinbrueche_korrigiert_df["anzahl_leistungseinbrueche"]
    / leistungseinbrueche_korrigiert_df["messungen_bei_hoher_einstrahlung"]
    * 100
).round(2)

leistungseinbrueche_korrigiert_df
```
