# Analyse der Leistung von zwei Photovoltaikanlagen

## Projektziel

In diesem Projekt werden die Produktions- und Wetterdaten von zwei Photovoltaikanlagen untersucht.

Ziel ist es herauszufinden, wie Sonneneinstrahlung und Temperatur die Stromproduktion beeinflussen. Außerdem wird untersucht, ob einzelne Wechselrichter oder Zeiträume ungewöhnlich niedrige Leistungen aufweisen.

## Hauptfrage

Wie beeinflussen Sonneneinstrahlung und Temperatur die Stromproduktion der beiden PV-Anlagen, und lassen sich auffällige Leistungseinbußen erkennen?

## Hypothesen

1. Je höher die Sonneneinstrahlung ist, desto höher ist die erzeugte AC-Leistung.
2. Die Modultemperatur hängt stärker mit der Stromproduktion zusammen als die Außentemperatur.
3. Einzelne Wechselrichter erzeugen weniger Energie als die übrigen Wechselrichter derselben Anlage.
4. Anlage 2 weist häufiger auffällige Leistungseinbrüche auf als Anlage 1.
5. Die höchste Stromproduktion wird ungefähr zur Mittagszeit erreicht.

# Analyse:

## Analyse 1: Einfluss der Sonneneinstrahlung auf die Stromproduktion

### Hypothese

Je höher die Sonneneinstrahlung ist, desto höher ist die erzeugte AC-Leistung.

## Korrelation für Anlage 1 berechnen

```python
# Daten von Anlage 1 auswählen

anlage1_zeitpunkt = zeitpunkt_df[zeitpunkt_df["anlage"] == "Anlage 1"]
```

```python
# Zusammenhang zwischen Sonneneinstrahlung und AC-Leistung berechnen

korrelation_anlage1 = anlage1_zeitpunkt[
    ["sonneneinstrahlung", "ac_leistung"]
].corr()

korrelation_anlage1
```

## Korrelation für Anlage 2 berechnen

```python
# Daten von Anlage 2 auswählen

anlage2_zeitpunkt = zeitpunkt_df[zeitpunkt_df["anlage"] == "Anlage 2"]
```

```python
# Zusammenhang zwischen Sonneneinstrahlung und AC-Leistung berechnen

korrelation_anlage2 = anlage2_zeitpunkt[["sonneneinstrahlung", "ac_leistung"]].corr()

korrelation_anlage2
```

## Beide Werte übersichtlich ausgeben

```python
# Korrelationswerte beider Anlagen anzeigen

print("Korrelation Anlage 1:",round(korrelation_anlage1.loc["sonneneinstrahlung", "ac_leistung"], 2))

print("Korrelation Anlage 2:",round(korrelation_anlage2.loc["sonneneinstrahlung", "ac_leistung"], 2))
```

## Streudiagramm erstellen

```python
# Zusammenhang zwischen Sonneneinstrahlung und AC-Leistung darstellen

fig = px.scatter(
    zeitpunkt_df,
    x="sonneneinstrahlung",
    y="ac_leistung",
    color="anlage",
    opacity=0.5,
    title="Sonneneinstrahlung beeinflusst die AC-Leistung beider Anlagen stark",
    labels={
        "sonneneinstrahlung": "Sonneneinstrahlung",
        "ac_leistung": "AC-Leistung",
        "anlage": "PV-Anlage"
    }
)

fig.show()
```

### Ergebnis

Bei beiden Anlagen besteht ein sehr starker positiver Zusammenhang zwischen Sonneneinstrahlung und AC-Leistung.

Die Korrelation beträgt bei Anlage 1 rund 1,00 und bei Anlage 2 rund 0,91. Mit steigender Sonneneinstrahlung steigt somit auch die erzeugte AC-Leistung.

Bei Anlage 2 ist jedoch eine stärkere Streuung zu erkennen. Mehrere Messpunkte zeigen trotz hoher Sonneneinstrahlung eine ungewöhnlich niedrige AC-Leistung. Dies kann auf zeitweise Leistungseinbrüche oder technische Auffälligkeiten hinweisen.

Die Hypothese wird bestätigt.

```python
print("Korrelation Anlage 1:",round(korrelation_anlage1.loc["sonneneinstrahlung", "ac_leistung"], 3))

print("Korrelation Anlage 2:",round(korrelation_anlage2.loc["sonneneinstrahlung", "ac_leistung"], 3))
```

## Analyse 2: Temperatur und Stromproduktion

### Hypothese

Die Modultemperatur hängt stärker mit der AC-Leistung zusammen als die Außentemperatur.

## Korrelationen für Anlage 1 berechnen

```python
# Korrelationen zwischen Wetter und AC-Leistung für Anlage 1

korrelation_temp_anlage1 = anlage1_zeitpunkt[["ac_leistung","aussentemperatur","modultemperatur","sonneneinstrahlung"]].corr().round(2)

korrelation_temp_anlage1
```

## Korrelationen für Anlage 2 berechnen

```python
# Korrelationen zwischen Wetter und AC-Leistung für Anlage 2

korrelation_temp_anlage2 = anlage2_zeitpunkt[[
    "ac_leistung",
    "aussentemperatur",
    "modultemperatur",
    "sonneneinstrahlung"
]].corr().round(2)

korrelation_temp_anlage2
```

## Relevante Werte direkt vergleichen

```python
# Temperatur-Korrelationen beider Anlagen übersichtlich ausgeben
print("Temperatur-Korrelationen:")
print("Anlage 1 - Außentemperatur:",korrelation_temp_anlage1.loc["ac_leistung", "aussentemperatur"])

print("Anlage 1 - Modultemperatur:",korrelation_temp_anlage1.loc["ac_leistung", "modultemperatur"])

print("Anlage 2 - Außentemperatur:",korrelation_temp_anlage2.loc["ac_leistung", "aussentemperatur"])

print("Anlage 2 - Modultemperatur:",korrelation_temp_anlage2.loc["ac_leistung", "modultemperatur"])
```

## Heatmap für Anlage 1

```python
# Korrelationen von Anlage 1 als Heatmap darstellen

fig = px.imshow(
    korrelation_temp_anlage1,
    text_auto=True,
    aspect="auto",
    color_continuous_scale="RdBu_r",
    zmin=-1,
    zmax=1,
    title="Zusammenhänge zwischen Wetter und AC-Leistung - Anlage 1"
)

fig.show()
```

## Heatmap für Anlage 2

```python
# Korrelationen von Anlage 2 als Heatmap darstellen

fig = px.imshow(
    korrelation_temp_anlage2,
    text_auto=True,
    aspect="auto",
    color_continuous_scale="RdBu",
    zmin=-1,
    zmax=1,
    title="Zusammenhänge zwischen Wetter und AC-Leistung - Anlage 2",
)

fig.show()
```

### Ergebnis

Bei beiden PV-Anlagen besteht zwischen der Modultemperatur und der AC-Leistung ein stärkerer Zusammenhang als zwischen der Außentemperatur und der AC-Leistung.

Bei Anlage 1 beträgt die Korrelation zwischen Modultemperatur und AC-Leistung 0,96. Die Korrelation mit der Außentemperatur liegt dagegen nur bei 0,73.

Bei Anlage 2 beträgt die Korrelation mit der Modultemperatur 0,87 und mit der Außentemperatur 0,65.

Die Hypothese wird daher bestätigt.

Die Ergebnisse bedeuten jedoch nicht, dass eine höhere Modultemperatur automatisch eine höhere Leistung verursacht. Die Module werden bei stärkerer Sonneneinstrahlung gleichzeitig wärmer und erzeugen mehr Strom. Die Sonneneinstrahlung ist deshalb weiterhin der wichtigste untersuchte Einflussfaktor.

## Analyse 3: Ertrag der einzelnen Wechselrichter

### Hypothese

Einzelne Wechselrichter erzeugen im untersuchten Zeitraum weniger Energie als die übrigen Wechselrichter derselben Anlage.

## Tagesertrag je Wechselrichter bestimmen

```python
# Höchsten Tagesertrag jedes Wechselrichters 

wechselrichter_tag_df = analyse_df.groupby(
    ["anlage", "tag", "wechselrichter_id"],
    as_index=False
)["tagesertrag"].max()

wechselrichter_tag_df.head()
```

```python
# Tageserträge für den gesamten Zeitraum 

wechselrichter_ertrag_df = wechselrichter_tag_df.groupby(["anlage", "wechselrichter_id"],as_index=False)["tagesertrag"].sum()
```

```python
# Ergebnisspalte umbenennen

wechselrichter_ertrag_df = wechselrichter_ertrag_df.rename(columns={"tagesertrag": "ertrag_gesamt"})

wechselrichter_ertrag_df.head()
```

## Durchschnitt je Anlage berechnen

```python
# Durchschnittlichen Wechselrichter-Ertrag jeder Anlage ergänzen

wechselrichter_ertrag_df["durchschnitt_anlage"] = (wechselrichter_ertrag_df.groupby("anlage")["ertrag_gesamt"].transform("mean"))

wechselrichter_ertrag_df.head()
```

## Abweichung vom Durchschnitt berechnen

```python
# Prozentuale Abweichung vom Durchschnitt berechnen

wechselrichter_ertrag_df["abweichung_prozent"] = (
    (wechselrichter_ertrag_df["ertrag_gesamt"]- wechselrichter_ertrag_df["durchschnitt_anlage"])
    / wechselrichter_ertrag_df["durchschnitt_anlage"]* 100).round(2)


print("Beispiel zur Bedeutung:")
print("")
print("0 = genau im Durchschnitt")
print("")
print("-5 = 5 % unter dem Durchschnit" )
print("")
print("8 = 8 % über dem Durchschnitt")

wechselrichter_ertrag_df.head()
```

## Schwächste Wechselrichter anzeigen

```python
# Wechselrichter nach Anlage und Ertrag sortieren

wechselrichter_ertrag_sortiert = wechselrichter_ertrag_df.sort_values(["anlage", "ertrag_gesamt"])

# Die drei schwächsten Wechselrichter jeder Anlage anzeigen

schwaechste_wechselrichter = wechselrichter_ertrag_sortiert.groupby("anlage").head(3)

schwaechste_wechselrichter.round({"ertrag_gesamt": 2,"durchschnitt_anlage": 2,"abweichung_prozent": 2})
```

## Analyse 4: Auffällige Leistungseinbrüche

### Hypothese

Anlage 2 weist häufiger auffällige Leistungseinbrüche auf als Anlage 1.

Als auffällig gilt in dieser Analyse ein Messzeitpunkt, bei dem die Sonneneinstrahlung mindestens 60 % des Höchstwertes der jeweiligen Anlage erreicht, während die AC-Leistung unter 30 % ihres Höchstwertes liegt.

## Werte normalisieren

```python
# AC-Leistung relativ zum Höchstwert jeder Anlage

zeitpunkt_df["ac_leistung_relativ"] = (zeitpunkt_df["ac_leistung"]/ zeitpunkt_df.groupby("anlage")["ac_leistung"].transform("max"))

zeitpunkt_df.head(3)
```

```python
# Sonneneinstrahlung relativ zum Höchstwert jeder Anlage berechnen

zeitpunkt_df["einstrahlung_relativ"] = (zeitpunkt_df["sonneneinstrahlung"] / zeitpunkt_df.groupby("anlage")["sonneneinstrahlung"].transform("max"))

zeitpunkt_df.head(3)
```

```python
# kontrolle
zeitpunkt_df[["anlage","ac_leistung","ac_leistung_relativ","sonneneinstrahlung","einstrahlung_relativ"]].head(3)
```

## Zeitpunkte mit hoher Einstrahlung auswählen


```python
# Nur Messungen mit mindestens 60 % der maximalen Einstrahlung auswählen

hohe_einstrahlung_df = zeitpunkt_df[zeitpunkt_df["einstrahlung_relativ"] >= 0.60].copy()

hohe_einstrahlung_df.head()
```

## Leistungseinbrüche markieren

```python
# Leistung unter 30 % trotz hoher Einstrahlung als auffällig markieren

hohe_einstrahlung_df["leistungseinbruch"] = (hohe_einstrahlung_df["ac_leistung_relativ"] < 0.30)
```

```python
# kontrolle

hohe_einstrahlung_df[["anlage","datum","einstrahlung_relativ","ac_leistung_relativ","leistungseinbruch"]].head()
```

### Hypothese

Die höchste Stromproduktion wird bei beiden PV-Anlagen ungefähr zur Mittagszeit erreicht.

# Fazit

Die Sonneneinstrahlung ist der wichtigste untersuchte Einflussfaktor auf die Stromproduktion beider PV-Anlagen. Mit zunehmender Sonneneinstrahlung steigt normalerweise auch die erzeugte AC-Leistung.

Anlage 1 zeigt insgesamt einen gleichmäßigeren Zusammenhang zwischen Sonneneinstrahlung und Stromproduktion. Auch die Erträge ihrer Wechselrichter liegen überwiegend nah beieinander.

Anlage 2 weist dagegen deutlich größere Schwankungen auf. Mehrere Wechselrichter erzeugen wesentlich weniger Energie als der Anlagendurchschnitt. Zusätzlich wurden mehrere Zeiträume erkannt, in denen die AC-Leistung trotz hoher Sonneneinstrahlung ungewöhnlich niedrig war.

Die Hauptfrage kann daher so beantwortet werden: Sonneneinstrahlung und Modultemperatur hängen stark mit der Stromproduktion zusammen. Gleichzeitig lassen sich bei Anlage 2 auffällige Leistungseinbußen und Unterschiede zwischen einzelnen Wechselrichtern erkennen.
